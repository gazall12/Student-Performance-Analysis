# Student Performance Analysis

The objective of this project is to analyze student performance data using Python. The project includes data loading, cleaning, analysis, and visualization to gain insights into student academic performance.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Step 1: Load Dataset

```python
df = pd.read_csv('student-mat.csv', sep=';')
#display first 5rows
df.head()
```

## Step 2: Explore Dataset

Understanding the dataset structure, dimensions, data types, and checking for missing values.

```python
# Dataset shape
print("Dataset Shape:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\n")
print("Data Types:")
print(df.dtypes)

print("\n")
print("Missing Values:")
print(df.isnull().sum())
```

### Observation

No missing values were found in the dataset.

## Step 3: Remove Duplicate Records


```python
duplicates = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicates)

df = df.drop_duplicates()

print("Shape After Removing Duplicates:", df.shape)
```

## Step 4: Data Analysis

```python
average_grade = df['G3'].mean()

print("Average Final Grade (G3):", round(average_grade,2))
```

```python
above_15 = (df['G3'] > 15).sum()

print("Students Scoring Above 15:", above_15)
```

```python
correlation = df['studytime'].corr(df['G3'])

print("Correlation between Study Time and Final Grade:", round(correlation,2))
```

```python
gender_performance = df.groupby('sex')['G3'].mean()

print(gender_performance)
```

## Step 5: Data Visualization

```python
plt.figure(figsize=(8,5))

plt.hist(df['G3'], bins=10)

plt.title('Distribution of Final Grades')
plt.xlabel('Final Grade (G3)')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df['studytime'], df['G3'])

plt.title('Study Time vs Final Grade')
plt.xlabel('Study Time')
plt.ylabel('Final Grade (G3)')

plt.show()
```

```python
gender_avg = df.groupby('sex')['G3'].mean()

gender_avg.plot(kind='bar')

plt.title('Average Final Grade by Gender')
plt.xlabel('Gender')
plt.ylabel('Average Grade')

plt.show()
```
